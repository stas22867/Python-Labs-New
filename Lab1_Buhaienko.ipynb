```python
def task_4(user_input=None):
    """
    Реалізація сортування методом бульбашки.
    Підраховує кількість виконаних перестановок.
    Вхідні дані передаються як параметр або вводяться через input().
    """
    if user_input is None:
        user_input = input()

    parts = user_input.strip().split()
    if not parts:
        return "Помилка. Введено порожній рядок"

    # Контроль значень: перевірка, чи всі введені елементи є цілими числами
    arr = []
    for item in parts:
        try:
            arr.append(int(item))
        except ValueError:
            return "Помилка. У списку присутнє нечислове значення"

    # Сортування методом бульбашки з підрахунком перестановок
    n = len(arr)
    swaps_count = 0
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps_count += 1

    return f"Було виконано {swaps_count} переставлень"
```

```python
# Тестовий приклад 1 із завдання
task_4("5 6 23 8 12 5")
```

```python
# Тестовий приклад 2 із завдання
task_4("7 3 1 7 4 2 8 2")
```

```python
# Тест контролю значень (як на сторінці 1 методички)
task_4("3 4 23 5 23 str 3 4")
```
