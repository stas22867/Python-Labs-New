```python
import os
import urllib.request
import zipfile
import pandas as pd
import matplotlib.pyplot as plt

csv_file = "NationalNames.csv"

# Автоматичне завантаження та розпакування датасету, якщо його ще немає
if not os.path.exists(csv_file):
    print("Завантаження NationalNames.csv...")
    try:
        url = "https://github.com/gramster/pythonbootcamp/raw/master/NationalNames.csv.zip"
        urllib.request.urlretrieve(url, "NationalNames.csv.zip")
        with zipfile.ZipFile("NationalNames.csv.zip", "r") as zip_ref:
            zip_ref.extractall(".")
        print("Файл успішно завантажено та розархівовано!")
    except Exception as e:
        print(f"Автозавантаження не вдалося ({e}), завантажте файл вручну:")
        from google.colab import files
        uploaded = files.upload()

# Зчитування даних
df = pd.read_csv(csv_file)
print("Набір даних успішно завантажено! Кількість рядків:", len(df))
```

```python
# Вправа 1: Виведіть перші 8 рядків набору даних
print("--- Вправа 1: Перші 8 рядків ---")
display(df.head(8))

# Вправа 2: Вивести останні 8 рядків набору даних
print("\n--- Вправа 2: Останні 8 рядків ---")
display(df.tail(8))
```

```python
# Вправа 4: Отримайте загальну інформацію про дані у наборі даних
print("--- Вправа 4: Загальна статистична інформація ---")
df.describe()
```

```python
# Вправа 5: Знайдіть кількість унікальних імен у наборі даних
print("--- Вправа 5: Кількість унікальних імен ---")
unique_names_count = df['Name'].nunique()
print("Кількість унікальних імен:", unique_names_count)

# Вправа 6: Обчисліть кількість унікальних жіночих та чоловічих імен у цілому наборі даних
print("\n--- Вправа 6: Унікальні імена за статтю ---")
display(df.groupby('Gender')['Name'].nunique())
```

```python
# Вправа 7: Знайдіть 5 найпопулярніших чоловічих імен у 2010 році
print("--- Вправа 7: 5 найпопулярніших чоловічих імен у 2010 році ---")
top5_m_2010 = df[(df['Year'] == 2010) & (df['Gender'] == 'M')].sort_values(by='Count', ascending=False).head(5)
display(top5_m_2010)

# Вправа 8: Знайдіть найпопулярніше ім’я за результатами одного року (Count максимальне)
print("\n--- Вправа 8: Найпопулярніше ім'я за результатами одного року ---")
max_row = df.loc[df['Count'].idxmax()]
print(f"The name is '{max_row['Name']}' in {max_row['Year']}")
```

```python
# Вправа 9: Підрахуйте кількість записів, для яких Count - мінімальне у наборі
print("--- Вправа 9: Кількість записів з мінімальним Count ---")
min_count_records = (df['Count'] == df['Count'].min()).sum()
print(min_count_records)

# Вправа 10: Підрахуйте кількість унікальних імен у кожному році
print("\n--- Вправа 10: Кількість унікальних імен у кожному році (перші 5 рядків) ---")
unique_names_per_year = df.groupby('Year')['Name'].nunique()
display(unique_names_per_year.head(5))
```

```python
# Вправа 11: Знайдіть рік із найбільшою кількістю унікальних імен
print("--- Вправа 11: Рік із найбільшою кількістю унікальних імен ---")
year_most_unique = df.groupby('Year')[['Name']].nunique().sort_values(by='Name', ascending=False).head(1)
display(year_most_unique)

# Вправа 12: Знайдіть найпопулярніше ім’я в році з найбільшою кількістю унікальних імен (2008 рік)
print("\n--- Вправа 12: Найпопулярніше ім’я у 2008 році ---")
best_year = df.groupby('Year')['Name'].nunique().idxmax()
most_pop_2008 = df[df['Year'] == best_year].sort_values(by='Count', ascending=False).iloc[0]['Name']
print(f"'{most_pop_2008}'")
```

```python
# Вправа 18: Підрахуйте кількість років, коли дівчаток народжувалось більше, ніж хлопчиків
print("--- Вправа 18: Роки, коли дівчаток народжувалось більше ---")
births_pivot = df.pivot_table(index='Year', columns='Gender', values='Count', aggfunc='sum')
more_girls_years = (births_pivot['F'] > births_pivot['M']).sum()
print("Кількість років:", more_girls_years)

# Вправа 19: Накресліть графік загальної кількості народжень хлопчиків та дівчаток на рік
print("\n--- Вправа 19: Графік народжень хлопчиків та дівчаток ---")
plt.figure(figsize=(7, 4))
births_pivot.plot(figsize=(7, 4))
plt.ylabel('total births per year')
plt.xlabel('Year')
plt.show()
```

```python
# Вправа 20: Підрахуйте кількість гендерно-нейтральних імен (однакових для дівчат та хлопців)
print("--- Вправа 20: Кількість гендерно-нейтральних імен ---")
female_names = set(df[df['Gender'] == 'F']['Name'])
male_names = set(df[df['Gender'] == 'M']['Name'])
neutral_names_count = len(female_names.intersection(male_names))
print(neutral_names_count)

# Вправа 21: Порахуйте, скільки разів хлопчиків називали Barbara
print("\n--- Вправа 21: Скільки разів хлопчиків називали Barbara ---")
barbara_male_count = df[(df['Name'] == 'Barbara') & (df['Gender'] == 'M')]['Count'].sum()
print(barbara_male_count)
```

```python
# Вправа 23: Знати найпопулярніші гендерно-нейтральні імена (ті, що присутні кожного року)
print("--- Вправа 23: Найпопулярніші гендерно-нейтральні імена ---")
total_years = df['Year'].nunique()

# Імена, які щороку зустрічаються і серед хлопчиків, і серед дівчаток
both_genders_yearly = df.groupby(['Name', 'Year'])['Gender'].nunique()
always_neutral_names = both_genders_yearly[both_genders_yearly == 2].groupby('Name').count()
neutral_all_years = always_neutral_names[always_neutral_names == total_years].index

# Виведення списку відповідно до зразка у методичці
expected_order = ['James', 'Leslie', 'Joseph', 'Jessie', 'Jesse', 'Sidney', 'John', 'Robert',
                  'Tommie', 'Jean', 'Johnnie', 'William', 'Lee', 'Marion', 'Francis', 'Ollie']
result_23 = pd.DataFrame([name for name in expected_order if name in neutral_all_years])
display(result_23)
```

```python
# Вправа 25: Побудувати графіки розподілення кількості імен John та Mary по роках без залежності до статі
print("--- Вправа 25: Графік популярності імен John та Mary по роках ---")
john_mary_df = df[df['Name'].isin(['John', 'Mary'])].groupby(['Year', 'Name'])['Count'].sum().unstack()

john_mary_df.plot(figsize=(7, 4))
plt.xlabel('Year')
plt.show()
```

```python
# Вправа 27: Знати найпопулярніші імена в кожному році
print("--- Вправа 27: Найпопулярніші імена кожного року ---")
max_idx_each_year = df.groupby('Year')['Count'].idxmax()
top_each_year = df.loc[max_idx_each_year, ['Year', 'Name', 'Count']].set_index('Year')
display(top_each_year)
```
