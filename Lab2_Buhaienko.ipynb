## Завдання 1
**Постановка задачі:**  
Введіть масив цілих чисел, що складається з $n > 2$ різних елементів. Поміняйте місцями максимальний і мінімальний елементи масиву. $n$ слід вводити з клавіатури.

```python
def swap_min_max(arr):
    """
    Функція міняє місцями мінімальний та максимальний елементи масиву.
    Повертає новий масив із переставленими елементами.
    """
    if len(arr) <= 2:
        return "Помилка: кількість елементів має бути більшою за 2"

    # Робимо копію, щоб не змінювати вихідний список
    result = list(arr)

    min_idx = result.index(min(result))
    max_idx = result.index(max(result))

    # Обмін місцями
    result[min_idx], result[max_idx] = result[max_idx], result[min_idx]
    return result

# Інтерактивна функція для введення з клавіатури за вимогою завдання
def task_1_input():
    n = int(input("Введіть n (n > 2): "))
    elements = [int(x) for x in input(f"Введіть {n} чисел через пробіл: ").split()]
    print("Результат:", swap_min_max(elements))
```

```python
# Тестові приклади з методички до Завдання 1
print("Тест 1 (4, [1, 4, 2, 5]):        ", swap_min_max([1, 4, 2, 5]))
print("Тест 2 (3, [2, 6, 8]):           ", swap_min_max([2, 6, 8]))
print("Тест 3 (5, [5, 25, 30, 20, 15]): ", swap_min_max([5, 25, 30, 20, 15]))
```

## Завдання 2
**Постановка задачі:**  
Надано список цілих чисел, видаліть найменше значення в ньому. **Не змінюйте початковий список.**  
Якщо є кілька елементів з однаковим значенням, видаліть той, що із найменшим індексом.  
Якщо ви отримуєте порожній список, поверніть порожній список.  
Не змінюйте порядок елементів, які залишилися.

```python
def remove_smallest(lst):
    """
    Видаляє перший знайдений найменший елемент списку.
    Не змінює початковий список.
    """
    if not lst:
        return []

    # Створюємо копію вихідного списку
    result = list(lst)
    min_value = min(result)

    # Метод .remove() видаляє перше входження елемента (тобто з найменшим індексом)
    result.remove(min_value)
    return result
```

```python
# Тестові приклади з методички до Завдання 2
test_orig = [2, 2, 1, 2, 1]

print("Тест 1 [1, 2, 3, 4, 5]: ", remove_smallest([1, 2, 3, 4, 5]))
print("Тест 2 [5, 3, 2, 1, 4]: ", remove_smallest([5, 3, 2, 1, 4]))
print("Тест 3 [2, 2, 1, 2, 1]: ", remove_smallest(test_orig))
print("Тест 4 порожній []:    ", remove_smallest([]))

# Перевірка, що початковий список не змінився
print("Перевірка незмінності початкового списку:", test_orig)
```

## Завдання 3
**Постановка задачі:**  
У заданому списку перший елемент має стати останнім.  
Порожній список або список лише з одним елементом має залишатися незмінним.

```python
def move_first_to_last(lst):
    """
    Переміщує перший елемент у кінець списку.
    Для порожнього або списку з 1 елементом повертає список без змін.
    """
    if len(lst) <= 1:
        return list(lst)

    # Зріз: починаючи з 1-го індексу + перший елемент списку в кінці
    return lst[1:] + [lst[0]]
```

```python
# Тестові приклади з методички до Завдання 3
print("Тест 1 [1, 2, 3, 4]: ", move_first_to_last([1, 2, 3, 4]))
print("Тест 2 [1]:          ", move_first_to_last([1]))
print("Тест 3 []:           ", move_first_to_last([]))
```
